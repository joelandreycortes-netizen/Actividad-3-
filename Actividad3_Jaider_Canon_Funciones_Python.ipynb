### Función list()

**Descripción**

La función `list()` permite crear una lista o convertir otro tipo de dato iterable (como una cadena, una tupla o un rango) en una lista.

```python
texto = "Python"
letras = list(texto)
print(letras)

numeros = list(range(1, 6))
print(numeros)
```

**Explicación**

La función `list()` toma un iterable y devuelve una lista con sus elementos. En el primer ejemplo, la cadena "Python" se separa en una lista de letras. En el segundo, `range(1, 6)` se convierte en la lista `[1, 2, 3, 4, 5]`.

### Función dict()

**Descripción**

La función `dict()` permite crear un diccionario, una estructura que almacena datos en pares de clave y valor.

```python
estudiante = dict(nombre="Jaider", edad=21, carrera="Ciencia de Datos")
print(estudiante)
print(estudiante["nombre"])

pares = [("a", 1), ("b", 2)]
print(dict(pares))
```

**Explicación**

La función `dict()` crea un diccionario a partir de argumentos con nombre o de una lista de pares clave-valor. Luego se puede acceder a cada valor usando su clave, por ejemplo `estudiante["nombre"]`.

### Función set()

**Descripción**

La función `set()` permite crear un conjunto, es decir, una colección de elementos únicos y sin orden definido.

```python
numeros = [1, 2, 2, 3, 3, 3, 4]
unicos = set(numeros)
print(unicos)
```

**Explicación**

La función `set()` elimina los elementos repetidos de un iterable. En este ejemplo, la lista tiene valores duplicados, pero el conjunto resultante solo conserva `{1, 2, 3, 4}`.

### Función tuple()

**Descripción**

La función `tuple()` permite crear una tupla, una secuencia ordenada de elementos que no se puede modificar (inmutable).

```python
frutas = ["Manzana", "Pera", "Uva"]
tupla_frutas = tuple(frutas)
print(tupla_frutas)
print(type(tupla_frutas))
```

**Explicación**

La función `tuple()` convierte un iterable en una tupla. A diferencia de las listas, las tuplas no permiten cambiar sus elementos una vez creadas, lo que es útil para proteger datos que no deben modificarse.

### Función all()

**Descripción**

La función `all()` devuelve `True` si todos los elementos de un iterable son verdaderos; de lo contrario devuelve `False`.

```python
notas = [4.0, 3.5, 5.0, 4.2]
print(all(n >= 3.0 for n in notas))

otras = [4.0, 2.5, 5.0]
print(all(n >= 3.0 for n in otras))
```

**Explicación**

La función `all()` revisa cada elemento y solo devuelve `True` si todos cumplen la condición. En el primer caso todas las notas son mayores o iguales a 3.0, por eso el resultado es `True`. En el segundo, hay una nota de 2.5, por lo tanto el resultado es `False`.

### Función any()

**Descripción**

La función `any()` devuelve `True` si al menos un elemento de un iterable es verdadero; si todos son falsos devuelve `False`.

```python
edades = [15, 17, 22, 16]
print(any(e >= 18 for e in edades))

menores = [12, 14, 16]
print(any(e >= 18 for e in menores))
```

**Explicación**

La función `any()` se detiene en cuanto encuentra un elemento verdadero. En la primera lista hay una persona de 22 años, por lo que el resultado es `True`. En la segunda no hay mayores de edad, así que el resultado es `False`.

### Función map()

**Descripción**

La función `map()` aplica una función a cada elemento de un iterable y devuelve un nuevo iterable con los resultados.

```python
numeros = [1, 2, 3, 4]
cuadrados = list(map(lambda x: x ** 2, numeros))
print(cuadrados)

textos = ["10", "20", "30"]
enteros = list(map(int, textos))
print(enteros)
```

**Explicación**

La función `map()` transforma cada elemento sin necesidad de escribir un ciclo. En el primer ejemplo se eleva al cuadrado cada número; en el segundo se convierte cada texto en un entero usando `int`. Se usa `list()` para visualizar el resultado.

### Función filter()

**Descripción**

La función `filter()` selecciona los elementos de un iterable que cumplen una condición determinada.

```python
numeros = [1, 2, 3, 4, 5, 6, 7, 8]
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(pares)
```

**Explicación**

La función `filter()` recibe una función que devuelve `True` o `False` y conserva solo los elementos para los que el resultado es `True`. En este ejemplo se obtienen únicamente los números pares de la lista.

### Función divmod()

**Descripción**

La función `divmod()` devuelve en una sola operación el cociente y el residuo de una división entera.

```python
cociente, residuo = divmod(17, 5)
print(cociente, residuo)
print(divmod(17, 5))
```

**Explicación**

La función `divmod(a, b)` devuelve una tupla con el resultado de `a // b` y de `a % b`. Al dividir 17 entre 5, el cociente es 3 y el residuo es 2.

### Función chr()

**Descripción**

La función `chr()` convierte un número entero en el carácter Unicode que le corresponde.

```python
print(chr(65))
print(chr(97))
print(chr(9731))
```

**Explicación**

La función `chr()` recibe un código Unicode y devuelve el carácter asociado. Por ejemplo, 65 corresponde a la letra "A", 97 a la "a" y 9731 al símbolo del muñeco de nieve.
