Introducción

Python es uno de los lenguajes de programación más utilizados en la actualidad debido a su facilidad de aprendizaje, sintaxis sencilla y amplia aplicación en diferentes áreas como el desarrollo de software, la automatización de procesos y la ciencia de datos.
En este trabajo se presentan varias funciones básicas de Python, describiendo su propósito, funcionamiento y aplicación mediante ejemplos prácticos. El objetivo es comprender cómo estas herramientas permiten resolver problemas de manera eficiente y facilitan el desarrollo de programas.
