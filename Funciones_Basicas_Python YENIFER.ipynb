Función min()
Descripción
La función min() permite encontrar el valor más pequeño dentro de un conjunto de datos.

```python
numeros = [8, 3, 12, 5]
print(min(numeros))
```

Explicación
La función analiza todos los elementos de la lista y devuelve el valor más pequeño. En este ejemplo el resultado es 3.
