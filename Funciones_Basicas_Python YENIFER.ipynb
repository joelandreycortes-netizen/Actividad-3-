Función min()
Descripción
La función min() permite encontrar el valor más pequeño dentro de un conjunto de datos.

```python
numeros = [8, 3, 12, 5]
print(min(numeros))
```

Explicación
La función analiza todos los elementos de la lista y devuelve el valor más pequeño. En este ejemplo el resultado es 3.

Función sum()
Descripción
La función sum() se utiliza para sumar todos los elementos numéricos de una lista.

```python
numeros = [10, 20, 30]

print(sum(numeros))
```

Explicación
La función sum() realiza la suma de todos los elementos de la lista. En este ejemplo, la suma de 10, 20 y 30 es igual a 60.

Función sorted()
Descripción
La función sorted() permite ordenar los elementos de una lista de menor a mayor.

```python
numeros = [5, 2, 8, 1]
print(sorted(numeros))
```

Explicación
La función sorted() organiza los elementos de una lista en orden ascendente. En este caso, los números se ordenan desde el menor hasta el mayor.

Función range()
Descripción
La función range() genera una secuencia de números dentro de un intervalo determinado.

```python
for i in range(1, 6):
  print(i)
```

Explicación
La función range(1, 6) genera los números desde 1 hasta 5. El número 6 no se incluye porque el límite superior no forma parte de la secuencia.

Función enumerate()
Descripción
La función enumerate() permite recorrer una lista asignando automáticamente un índice a cada elemento.

```python
frutas = ["Manzana", "Pera", "Uva"]

for indice, fruta in enumerate(frutas):
    print(indice, fruta)
```

Explicación
La función enumerate() agrega un número de posición a cada elemento de la lista. Esto facilita identificar la ubicación de cada elemento mientras se recorre la colección.

Función zip()
Descripción
La función zip() permite combinar elementos de dos o más listas en pares ordenados.

```python
nombres = ["Juan", "Ana"]
edades = [20, 22]

print(list(zip(nombres, edades)))
```

Explicación
La función zip() une los elementos de varias listas según su posición. En este ejemplo, cada nombre se asocia con la edad correspondiente.

Función reversed()
Descripción
La función reversed() permite invertir el orden de los elementos de una secuencia, como una lista.

```python
numeros = [1, 2, 3, 4]

print(list(reversed(numeros)))
```

Explicación
La función reversed() recorre los elementos de la lista desde el último hasta el primero. En este ejemplo, la lista original se muestra en orden inverso.

Función pow()
Descripción
La función pow() permite calcular la potencia de un número elevándolo a un exponente determinado.

```python
print(pow(2, 4))
```

Explicación
La función pow() eleva un número a una potencia específica. En este ejemplo, 2 elevado a la potencia 4 da como resultado 16.

Función isinstance()
Descripción
La función isinstance() permite verificar si una variable pertenece a un tipo de dato específico.

```python
x = 15
print(isinstance(x, int))
```

Explicación
La función isinstance() comprueba si una variable corresponde a un tipo de dato determinado. En este caso, la variable x es un número entero, por lo que el resultado es True.

Función bool()
Descripción
La función bool() permite convertir un valor en un resultado lógico, es decir, verdadero (True) o falso (False).

```python
print(bool(1))
print(bool(0))
```

Explicación
La función bool() evalúa un valor y determina si es verdadero o falso. En Python, el número 1 se interpreta como True y el número 0 como False.
