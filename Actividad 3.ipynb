#**Funciones de python **

## 1. Funcion print() - mostrar informacion
esta funcion Sirve para mostrar mensajes, resultados y datos en la pantalla.

Ejemplo:  
print (" hola , mundo ")

print( 5 + 3 )

Resultado: muestra Hola, mundo y después 8.

## 2. Funcion input ()- Recibir datos
Esta funcion nos permite que el usuario escriba información por el teclado.

Ejemplo : nombre = input("Escribe tu nombre: ")

print("Hola", nombre)

Uso práctico: pedir el nombre de un estudiante.


## 3. Funcion int() - convertir a numero entero
Esta funcion convierte un valor compatible en un número sin decimales.

Ejemplo : edad = int("18")

print(edad + 2)

Resultado: 20.

##4. Funcion float() - convertir a decimal
Esta funcion permite trabajar con números que tienen parte decimal.

Ejemplo : precio = float("4500.50")

print(precio)

Resultado: 4500.5

## 5. Funcion str() — Convertir a texto
Esta funcion convierte un valor en una cadena de texto.

Ejemplo :edad = 18

mensaje = "Tengo " + str(edad) + " años"

print(mensaje)

Resultado: Tengo 18 años.

##6. Funcion len() — Contar elementos
Esta funcion cuenta la cantidad de caracteres de un texto o elementos de una lista.

Ejemplo : nombre = "Carlos"

print(len(nombre))

Resultado: 6.

##7. Funcion type() — Identificar el tipo de dato
Esta funcion indica si un valor es entero, texto, decimal u otro tipo.

Ejemplo : numero = 25

print(type(numero))

Resultado: int, que significa número entero.

##8. Funcion abs() — Obtener el valor absoluto
Esta funcion devuelve la distancia de un número al cero, sin considerar su signo.

Ejemplo : print(abs(-15))

Resultado: 15.

##9. Funcion round() — Redondear números
Esta funcion permite aproximar un número decimal.

Ejemplo : print(round(3.786, 2))

Resultado: 3.79.

##10. Funcion max() — Encontrar el valor mayor
Esta funcion Devuelve el número más grande de un grupo de valores.

Ejemplo : print(max(12, 30, 8, 25))

Resultado: 30.
